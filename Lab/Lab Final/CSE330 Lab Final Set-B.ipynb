# Question02

```python

def calc_lagrange_basis(x, eval_x):
    resultArr = []
    result = 1
    for i in range(len(x)):
        for j in range(len(x)):
            if i != j:
                result *= ((2*(x[i]) - x[j]) / (3*(x[i]) - eval_x))
        resultArr.append(result)
        result = 1
    return resultArr

def Lagrange(y, Lagrange_Basis):
    result = 0
    for i in range(len(y)):
        result += y[i] * Lagrange_Basis[i]
    return result

# Driver Code
x=[0,3,5,7]
y=[0,20,90,85]
eval_x=2

Lagrange_Basis= calc_lagrange_basis(x,eval_x)
print(f"Lagrange Basis: {Lagrange_Basis}")

P_x=Lagrange(y,Lagrange_Basis)
print(f"\nInterpolated value P({eval_x}) = {P_x}")
```

# Question03

```python
f = Polynomial([7, 3, -5, 2])
f_prime = f.deriv(1)
x = 2
h = 0.4
tolerance = 0.0025
reductions = 0
actual_deriv = f_prime(x)

def forward_difference(f, x, h):
    return (f(x + h) - f(x)) / h

def error(f, f_prime, h, x, tolerance):
  reductions = 0
  while True:
    approx_deriv = forward_difference(f, x, h)
    error = abs(actual_deriv-approx_deriv)
    if error > tolerance:
        h /= 2
        reductions += 1
    else:
        break

  print(f"Minimum h: {h}")
  print(f"\nTotal number of reductions: {reductions}")
  print(f"\nAbsolute error: {error}")

#driver code
error(f,f_prime,h,x,tolerance)
```
